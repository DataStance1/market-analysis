```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("🩸 Halo Care Feminine Hygiene Survey .csv")
```

```python
df.columns
```

```python
responses = df['If you could tell a menstrual care brand one thing to improve your experience; what would it be?']
```

```python
type(responses)
```

```python
# Define themes and associated keywords for classification
themes_keywords = {
    "Comfort & Skin Sensitivity": ["comfortable", "comfort", "soft", "skin", "irritation", "sensitive", "abrasive", "rash", "reaction"],
    "Length, Width & Coverage": ["length", "long", "wider", "expand", "coverage", "sizeable", "roll over", "body types"],
    "Adhesive/Gum Quality (Stickiness)": ["gum", "stick", "adhesive", "wings", "hold", "glue", "panties", "underwear"],
    "Absorbency & Leak Protection": ["absorbent", "absorbency", "leak", "drip", "flow", "heavier", "soak", "protect"],
    "Affordability & Accessibility": ["affordable", "cheap", "price", "expensive", "accessible", "cost"],
    "Product Variety": ["night", "heavy flow", "liners", "variety", "options", "sizes", "different types"],
    "Sustainability & Materials": ["eco", "sustainable", "recyclable", "non-toxic", "organic", "materials", "fragrance-free"],
    "Pain Relief & Extra Features": ["pain", "relief", "menthol", "soothing", "cooling", "heating", "extra", "features", "educational"],
    "Positive Feedback / No Issues / Nil": ["nil", "keep up", "nothing", "good work"]
}
```

```python
# Function to classify each response
def classify_response(response, theme_keywords):
    response_lower = response.lower()
    matched_themes = []
    for theme, keywords in theme_keywords.items():
        if any(keyword in response_lower for keyword in keywords):
            matched_themes.append(theme)
    return matched_themes if matched_themes else ["Unclassified"]

# Classify all responses
classified_data = []
for resp in responses:
    themes = classify_response(resp, themes_keywords)
    classified_data.append({"Response": resp, "Themes": themes})
```

```python
# Create a DataFrame
df1 = pd.DataFrame(classified_data)
```

```python
df1
```

```python
# Expand themes into comma-separated string for readability
df1["Themes"] = df1["Themes"].apply(lambda x: "; ".join(x))

df1
```

```python
from collections import Counter

# Flatten the list of all themes
all_themes = sum((classify_response(resp, themes_keywords) for resp in responses), [])
theme_counts = Counter(all_themes)

# Convert to a DataFrame for easy display
theme_summary = pd.DataFrame(theme_counts.items(), columns=["Theme", "Count"]).sort_values(by="Count", ascending=False)
theme_summary
```
