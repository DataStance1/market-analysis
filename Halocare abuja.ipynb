```python
import pandas as pd
df = pd.read_csv('Product Audit Table—Feminine Hygiene Products (Abuja).csv')
```

```python
def classify_packaging_multi(text):
    if pd.isna(text):
        return 'Missing'

    text = text.lower()
    categories = []

    if any(k in text for k in ['yellow', 'purple', 'pink', 'colour', 'color', 'green', 'white', 'blue']):
        categories.append('Color-Focused')

    if any(k in text for k in ['portable', 'rectangular', 'compact', 'cubic', 'small', 'handy']):
        categories.append('Portable Design')

    if any(k in text for k in ['traditional', 'paper', 'carton']):
        categories.append('Traditional')

    if any(k in text for k in ['exotic', 'unique', 'fancy', 'attractive', 'night', 'mint', 'shower bag']):
        categories.append('Exotic or Unique')

    if any(k in text for k in ['feminine', 'fine', 'girly']):
        categories.append('Feminine Design')

    if any(k in text for k in ['besense', 'softcare', 'same as']):
        categories.append('Brand Comparison')

    if any(k in text for k in ['okay', 'good', 'nil', 'refer to earlier entries', 'same packaging', 'same as other']):
        categories.append('Unspecified/Generic')

    if 'different designs for different variants' in text:
        categories.append('Unspecified/Generic')

    return '; '.join(categories) if categories else 'Other'

# Apply the function to your packaging column
df['Packaging_Classifications'] = df['Packaging Description: What does the packaging look/feel like? '].apply(classify_packaging_multi)
```
