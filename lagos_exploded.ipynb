```python
import pandas as pd
```

```python
df = pd.read_excel('lagos_explode.xlsx')
```

```python
df.columns
```

```python
# Step 1: #removing leading and trailing spaces
df['Key Claims on Pack: E.g., leak protection, ultra-thin, odor control '] = df['Key Claims on Pack: E.g., leak protection, ultra-thin, odor control '].str.strip()
```

```python
# Step 1: #making all strings lower case
df['Key Claims on Pack: E.g., leak protection, ultra-thin, odor control '] = df['Key Claims on Pack: E.g., leak protection, ultra-thin, odor control '].str.lower()
```

```python
# Step 1: First, split the semicolon-separated themes into a list
df['Key Claims on Pack: E.g., leak protection, ultra-thin, odor control '] = df['Key Claims on Pack: E.g., leak protection, ultra-thin, odor control '].str.split(", ")
```

```python
df
```

```python
df = df.explode('Key Claims on Pack: E.g., leak protection, ultra-thin, odor control ').reset_index(drop=True)
df
```

```python
# Save to Excel
df.to_excel('lagos_exploded.xlsx', index=False)
```
